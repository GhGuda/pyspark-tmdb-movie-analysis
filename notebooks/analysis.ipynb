## Visual Analysis

The following plots explore relationships and trends
in movie performance metrics.


```python
import matplotlib.pyplot as plt
```

### Revenue vs Budget
This plot shows how movie budgets relate to revenue outcomes.


```python
plt.figure(figsize=(10,6))
plt.scatter(movies_df['budget_musd'], movies_df['revenue_musd'], alpha=0.7)
plt.title('Revenue vs Budget')
plt.xlabel('Budget (USD)')
plt.ylabel('Revenue (USD)')
plt.grid(True)
plt.show()
```

### ROI Distribution by Genre
This shows how return on investment varies across genres.



```python
df_genres = movies_df.copy()
df_genres["roi"] = (
    (df_genres["revenue_musd"] - df_genres["budget_musd"]) /
    df_genres["budget_musd"]
).replace([np.inf, -np.inf], np.nan)
df_genres = df_genres[df_genres["budget_musd"] >= 10]
df_genres['genres'] = df_genres['genres'].str.split('|')
df_genres = df_genres.explode('genres')

plt.figure(figsize=(12,6))
df_genres.boxplot(column='roi', by='genres', rot=90)
plt.title("ROI Distribution by Genre")
plt.suptitle("")
plt.xlabel("Genre")
plt.ylabel("ROI")
plt.show()
```

### Popularity vs Rating
Examines whether more popular movies are better rated.



```python
plt.figure(figsize=(10,6))
plt.scatter(movies_df['popularity'], movies_df['vote_average'])
plt.xlabel("Popularity")
plt.ylabel("Rating")
plt.title("Popularity vs Rating")
plt.grid(True)
plt.show()
```

### Yearly Box Office Trends
Shows total revenue trends over time.


```python
movies_df['year'] = movies_df['release_date'].dt.year
yearly = movies_df.groupby('year')['revenue_musd'].sum().reset_index()

plt.figure(figsize=(12,6))
plt.plot(yearly['year'], yearly['revenue_musd'])
plt.xlabel("Year")
plt.ylabel("Total Revenue")
plt.title("Yearly Box Office Revenue Trends")
plt.grid(True)
plt.show()
```

### Franchise vs Standalone Performance
Comparison of ROI between franchise and standalone movies.


```python
df_franchise = movies_df.copy()

df_franchise["roi"] = (
    (df_franchise["revenue_musd"] - df_franchise["budget_musd"]) /
    df_franchise["budget_musd"]
).replace([np.inf, -np.inf], np.nan)


df_franchise = df_franchise[df_franchise["budget_musd"] >= 10]

df_franchise["franchise_type"] = df_franchise["belongs_to_collection"].notna().map({
    True: "Franchise",
    False: "Standalone"
})

franchise_stats = df_franchise.groupby('franchise_type')[['roi']].mean()

plt.figure(figsize=(8,5))
franchise_stats['roi'].plot(kind='bar')
plt.title("ROI: Franchise vs Standalone")
plt.ylabel("ROI")
plt.grid(True)
plt.show()
```
